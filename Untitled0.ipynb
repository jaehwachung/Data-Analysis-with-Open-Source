```python
import requests

api_key = "7551586b5668616e3131325a6f525177"
datas = []
params = {
    'serviceKey': api_key,
    'returnType': 'json',
    'numOfRows': '100',
    'pageNo': '1',
    'sidoName': '서울',
    'ver': '1.0'
}

for year in range(2015,2025):
    for month in range(1,13):
        url = f"http://openapi.seoul.go.kr:8088/7551586b5668616e3131325a6f525177/json/energyUseDataSummaryInfo/1/5/{year}/{month:02d}"
        response = requests.get(url, params=params)

        if response.status_code == 200:
            print("api 호출 성공")
            print(response.json())
            data = response.json()
            datas.extend(data['energyUseDataSummaryInfo']['row'])
        else:
            print(f"API 호출 실패: {response.status_code}")
```
